```python
from pathlib import Path
import sys
import subprocess

# Escribe aquí la URL HTTPS del repositorio que contendrá los archivos de la Unidad 2.
REPO_URL = "https://github.com/bettypau-open-knowledge/del-dataset-a-unet_reto-2"
EN_COLAB = Path("/content").exists() and "google.colab" in sys.modules

if EN_COLAB:
    PROYECTO = Path("/content/reto_unidad_2")
    if (PROYECTO / ".git").exists():
        subprocess.run(["git", "-C", str(PROYECTO), "pull", "--ff-only"], check=True)
    elif PROYECTO.exists():
        raise RuntimeError(f"Existe {PROYECTO}, pero no es un repositorio Git.")
    else:
        if not REPO_URL.strip():
            raise ValueError("Configura REPO_URL con la URL HTTPS pública del repositorio de la Unidad 2.")
        subprocess.run(["git", "clone", REPO_URL, str(PROYECTO)], check=True)
else:
    PROYECTO = Path.cwd().resolve()
    if not (PROYECTO / "src").is_dir():
        PROYECTO = PROYECTO.parent

if not (PROYECTO / "src").is_dir():
    raise FileNotFoundError(f"No se encontró la carpeta src/ en {PROYECTO}.")
if str(PROYECTO) not in sys.path:
    sys.path.insert(0, str(PROYECTO))
print("Proyecto:", PROYECTO)
```
